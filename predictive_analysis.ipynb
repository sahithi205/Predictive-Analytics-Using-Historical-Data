```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
```

```python
data = pd.read_csv("data/sales_data.csv")

data.head()
```

```python
X = data[['Month']]
y = data['Sales']
```

```python
model = LinearRegression()

model.fit(X, y)
```

```python
predictions = model.predict(X)
```

```python
mae = mean_absolute_error(y, predictions)

print("Mean Absolute Error:", mae)
```

```python
future_months = pd.DataFrame({
    'Month': [37, 38, 39]
})

future_predictions = model.predict(future_months)

for month, sale in zip(future_months['Month'], future_predictions):
    print(f"Month {month}: {sale:.2f}")
```

```python
plt.figure(figsize=(8,5))

plt.scatter(X, y, label='Actual Sales')
plt.plot(X, predictions, label='Prediction Line')

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Sales Forecast Prediction")

plt.legend()

plt.show()
```
